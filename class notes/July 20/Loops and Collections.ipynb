
# 1. Data Collections
A data collection is a data type that allows us to store multiple values in a single variable.

Instead of creating many variables to store related data, Python lets us group them together in one collection.
## Why Do We Use Data Collections?

Data collections help us:
- Store multiple values together
- Organize related information
- Access data easily
- Loop through data efficiently
- Reduce repetitive code
- Perform operations like searching, sorting, counting, and updating

Without collections, programming large applications would be extremely difficult.

Python provides five major collection types: String, List, Tuple, Set,
and Dictionary.

```python
# String 
# A string stores text. Strings store characters inside quotation marks.
name = "Shristi"

print(name)
```

```python
#Lists
#A list stores multiple values in order.
fruits = ["Apple", "Banana", "Orange"]

print(fruits)
```

```python
# Tuple
# A tuple is similar to a list, but cannot be changed after it is created.
#A tuple is perfect for storing information that should remain constant.
#Examples

    #GPS coordinates
    #Months
    #Days of the week
    #RGB colors

coordinates = (15, 20)

print(coordinates)
```

```python
# Set
#A set stores only unique values.
numbers = {1, 2, 2, 3, 4}

print(numbers)
# here the duplicate 2 is removed automatically.
```

```python
#Dictionary
#A dictionary stores data as key-value pairs.
student = {
    "name": "Shristi",
    "age": 30,
    "major": "Marketing Analytics"
}

print(student)
```

## Loop
A loop is a programming structure that allows us to execute the same block of code repeatedly until a certain condition is met.

Instead of writing the same code multiple times, we use loops to automate repetitive tasks.

There are two types of Loops
 - For Loop
 - While Loop


## FOR Loop
A for loop is used to iterate over a sequence such as:

- List
- String
- Tuple
- Set
- Dictionary
- range()

```python
#printing student names
students = ["shristi", "deepika", "sindhu", "pawan"]

for student in students:
    print(student)
```

```python
#printing grocery list
shopping_list = ["Milk", "Eggs", "Bread", "Butter"]

for item in shopping_list:
    print("Buy:", item)
```

```python
#printing each character
name = "Shristi"

for letter in name:
    print(letter)
```

```python
#calculating total sales
sales = [250, 180, 300, 420]
total = 0

for sale in sales:
    total += sale

print("Total Sales: $", total)
```

```python
#multiplication table
number = 5

for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")
```

## While Loop
A while loop keeps running as long as a condition is True.

Unlike a for loop, a while loop does not know in advance how many times it will run.

```python
#counting numbers
count = 1

while count <= 5:
    print(count)
    count += 1
```

```python
password = ""

while password != "python123":
    password = input("Enter password: ")

print("Access Granted")
```

```python
stored_password = "python123"

entered_password = input("Enter Password: ")

while entered_password != stored_password:
    print("Wrong Password")
    entered_password = input("Try Again: ")

print("Login Successful")
```

## Range
The range() function is a built-in Python function that generates a sequence of numbers.

Instead of creating a list of numbers manually, range() generates them automatically.

It is most commonly used with for loops to control how many times the loop executes.

Syntax: range(start, stop)

```python
for number in range(3, 8): #here the range starts from 3 and stops at 7
    print(number)
```

```python
#sum of numbers
total = 0

for number in range(1, 6):
    total += number

print("Total:", total)
```

```python
#listing grocery items
items = ["Milk", "Bread", "Eggs", "Butter"]

for index in range(len(items)):
    print(index + 1, "-", items[index])
```

## Looping by Value VS Looping by Index

When working with collections like lists, tuples, or strings, we can loop through them in two different ways:

Loop by Value – Access the actual values stored in the collection.

Loop by Index – Access the position (index) of each value.

## Looping by Value

Looping by value means Python gives you each item directly from the collection.

You don't care about where the item is located—you only need the item itself.

```python
students = ["Deepika", "Prasuna", "Utsha", "Anna"]

for student in students:
    print(student)
```

## Looping by Index

Looping by index means Python gives you the position of each item instead of the item itself.

You then use the index to access the value.

```python
#printing student with index
students = ["Deepika", "Prasuna", "Utsha", "Anna"]

for index in range(len(students)):
    print(index, students[index])
```

```python
cities = ["Dayton", "Columbus", "Cincinnati"]

for i in range(len(cities)):
    print(i, cities[i])
```

## Reversing String in Python

Reversing a string means displaying the characters in the opposite order.

```python
# reversing name using slicing
name = "Shristi"

print(name[::-1])
```

## Reverse method in Python

The reverse() method reverses the order of items inside a list.

It changes the original list directly.

```python
fruits = ["Apple", "Banana", "Orange"]

print("Before:", fruits)

fruits.reverse()

print("After:", fruits)
```

```python
#countdown using reverse
countdown = [1, 2, 3, 4, 5]

countdown.reverse()

for number in countdown:
    print(number)

print("Start!")
```

## Break and Continue Statements
Both break and continue are used inside loops, but they do different jobs.

break stops the entire loop.

continue skips only the current iteration and moves to the next one.

## Break Statement
The break statement immediately ends the loop.

Once Python reaches break, it exits the loop and continues with the code written after the loop.

```python
# stop when number is found
numbers = [10, 20, 30, 40, 50]

for number in numbers:
    if number == 30:
        break

    print(number)
```

```python
#search for a product
products = ["Laptop", "Phone", "Watch", "Camera"]

for product in products:
    print("Checking:", product)

    if product == "Watch":
        print("Product found!")
        break
```

```python
#password verification
correct_password = "python123"

while True:
    entered_password = input("Enter password: ")

    if entered_password == correct_password:
        print("Access Granted")
        break

    print("Wrong Password")
```

## Continue Statement
The continue statement skips the remaining code in the current iteration and moves to the next iteration.

Unlike break, it does not stop the entire loop.

```python
#skip a product that is out of stock
products = ["Laptop", "Phone", "Watch", "Camera"]
out_of_stock = "Watch"

for product in products:
    if product == out_of_stock:
        continue

    print("Available:", product)
```

```python
#print odd numbers
for number in range(1, 11):
    if number % 2 == 0:
        continue

    print(number)
```

```python
#skip empty names
names = ["shristi", "", "abc", "", "def"]

for name in names:
    if name == "":
        continue

    print("Student:", name)
```

```python
#skip invalid marks
marks = [85, -10, 90, 120, 75]

for mark in marks:
    if mark < 0 or mark > 100:
        print("Invalid mark skipped:", mark)
        continue

    print("Valid mark:", mark)

```
