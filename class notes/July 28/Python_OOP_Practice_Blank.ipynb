# Encapsulation

Encapsulation means keeping the data (variables) and the methods (functions) that work with that data inside one class.

Instead of letting anyone change important data directly, we provide methods that control how the data is accessed or modified.

```python
# Example
class BankAccount: #Creates a new class

    def __init__(self, owner, balance): #creating method and it runs automatically when an object is created.
        self.owner = owner #Public attribute. it stores the owner name assigned
        self.__balance = balance #private attribute. it stores the starting balance

    def deposit(self, amount): #created deposit method
        self.__balance += amount #Take the current balance, add the deposit amount, and save the new balance.

    def show_balance(self): #created showblance method
        print("Current Balance:", self.__balance)


account = BankAccount("Shristi", 500)

account.deposit(200)

print(account.owner)
account.show_balance()
```

## Composition

Composition is an Object-Oriented Programming (OOP) concept where one class contains an object of another class.

Instead of making one class inherit from another (using inheritance), we simply create an object of another class inside our class and use its features.

## Composition is widely used because it helps organize programs into smaller, reusable pieces.
1. Makes large programs easier to manage - 
Instead of writing one huge class, we divide the program into smaller classes.
2. Promotes code reuse - 
Once a class is created, it can be reused in many other classes.
3. Reduces unnecessary inheritance
4. Easier to update - 
Updating one class is often enough, as the changes are automatically available to the classes that use it.

```python
class Battery:

    def charge(self):
        print("Battery is charging")


class Laptop:

    def __init__(self, brand):
        self.brand = brand
        self.battery = Battery()

    def charge_laptop(self):
        print(self.brand, "Laptop")
        self.battery.charge()


my_laptop = Laptop("Apple")

my_laptop.charge_laptop()
```

## Practice 3 - Dynamic Attributes

Normally, we create an object's attributes (variables) inside the __init__() method.

However, Python also allows us to add new attributes to an object after it has already been created.

These newly added attributes are called Dynamic Attributes.

## Why Dynamic Attributes are Useful
1. Makes objects more flexible -
You can add new information to an object whenever it is needed, even if it wasn't planned when the class was created.
2. Avoids creating unnecessary attributes - 
Instead of creating many attributes that every object may not need, you can add them only to the objects that require them.
Useful when data changes during runtime - 
Sometimes new information becomes available while the program is running. Dynamic attributes allow objects to store that new information without modifying the original class.
4. Saves development time -
You don't always need to rewrite the class whenever you want to store additional information for an object.


```python
class Student:

    def __init__(self, name):
        self.name = name

student = Student("Shristi")

student.course = "Marketing Analytics"
student.gpa = 4.0

print(student.name)
print(student.course)
print(student.gpa)
```

```python
class Car:

    def __init__(self, brand):
        self.brand = brand


car = Car("Toyota")

# Dynamic Attributes
car.color = "White"
car.owner = "Shristi"

print(car.brand)
print(car.color)
print(car.owner)
```

## Polymorphism

In Object-Oriented Programming (OOP), polymorphism means that different classes can have a method with the same name, but each class performs that method differently.
Polymorphism allows the same method name to behave differently depending on which object calls it.

There are two main types of polymorphism in Object-Oriented Programming.
1. Compile-Time Polymorphism 
2. Run-Time Polymorphism 


## Why Polymorphism is useful
1. Code Reusability -
The same method can be used by different objects without writing separate code.
2. Reduces Duplicate Code -
Instead of creating different method names for every class, different classes can use the same method name.
3. Easy to Extend -
New classes can be added without changing the existing code.


```python
class CreditCard:

    def pay(self):
        print("Payment made using Credit Card")


class PayPal:

    def pay(self):
        print("Payment made using PayPal")


class Cash:

    def pay(self):
        print("Payment made using Cash")


credit = CreditCard()
paypal = PayPal()
cash = Cash()

credit.pay()
paypal.pay()
cash.pay()
#Every class has the same method: pay(),But each object performs the payment differently.
```

## Complile-time Polymorphism
Compile-time polymorphism means the method to execute is decided before the program runs (during compilation).

It is commonly achieved using:
   1. Method Overloading
   2. Operator Overloading

```python
class Calculator:

    def add(self, a, b=0, c=0):
        return a + b + c


calc = Calculator()

print(calc.add(5))
print(calc.add(5, 6))
print(calc.add(5, 6, 7))
```

## Run-time Polymorphism
Run-time polymorphism means the method to execute is decided while the program is running.
Python mainly achieves this using Method Overriding.
Different classes define the same method, and Python chooses the correct one based on the object.

```python
class Manager:

    def work(self):
        print("Manager is planning the project")


class Developer:

    def work(self):
        print("Developer is writing code")


class Designer:

    def work(self):
        print("Designer is creating the user interface")


employees = [Manager(), Developer(), Designer()] #This line creates three different objects. All three objects are stored inside a list.

for employee in employees:
    employee.work()

#At the beginning of the loop, Python doesn't know which type of employee employee will refer to next.
#Because this decision is made during program execution (at runtime), this is called Run-Time Polymorphism.
```

## Method Overriding
Method Overriding occurs when a child class provides its own implementation of a method that already exists in the parent class.
The child class replaces the parent's method with its own version.
The method name remains the same, but the functionality changes.

## Why do we use Method Overriding?
Method Overriding allows child classes to customize or change the behavior inherited from the parent class without changing the parent class itself.

Benefits
Reuse the parent class code.
Allow each child class to have its own behavior.
Makes programs easier to maintain.
Supports Run-Time Polymorphism.

```python
Bonus = 2500
class Employee:

    def calculate_salary(self):
        print("Employee Base Salary: $3000")


class Manager(Employee):

    def calculate_salary(self):
        total_salary = 5000 + Bonus
        print(f"Base Salary: $5000")
        print(f"Bonus: ${Bonus}")
        print(f"Manager Total Salary: ${total_salary}")



class Intern(Employee):

    def calculate_salary(self):
        print("Intern Stipend: $1500")


employee = Employee()
manager = Manager()
intern = Intern()

employee.calculate_salary()
manager.calculate_salary()
intern.calculate_salary()
```

## Duck Typing
Duck Typing is a feature of Python where Python does not check an object's class or type.

Instead, Python only checks whether the object has the required method or behavior.

Duck Typing does not require inheritance.

The classes can be completely unrelated.

The only requirement is that they provide the same method.

```python
class Car:

    def start(self):
        print("Car engine started")


class Laptop:

    def start(self):
        print("Laptop is booting")


class WashingMachine:

    def start(self):
        print("Washing Machine started")


def power_on(device):
    device.start()


car = Car()
laptop = Laptop()
washing_machine = WashingMachine()

power_on(car)
power_on(laptop)
power_on(washing_machine)
```

## Class vs Instance Variables
* A Class Variable is a variable that belongs to the class itself, not to individual objects.

   This means: Every object shares the same value.
If the class variable changes, every object sees the updated value.

* An Instance Variable belongs to an individual object.
    
    Each object has its own copy. usually created inside __init__(). Changing one object's instance variable does not affect the others.

```python
class Product:

    # Class Variable
    store_name = "ABC Supermarket"

    def __init__(self, name, price):

        # Instance Variables
        self.name = name
        self.price = price


product1 = Product("Apple", 2)
product2 = Product("Bread", 4)

print(product1.store_name)
print(product2.store_name)

print(product1.name, product1.price)
print(product2.name, product2.price)

# Change the class variable
Product.store_name = "XYZ Mart"

print("\nAfter Change")
print(product1.store_name)
print(product2.store_name)
```

## Instance/Class/Static Methods

An Instance Method is a method that works with instance variables and receives self as its first parameter. It is called using an object.

A Class Method works with class variables and receives cls as its first parameter. It is created using the @classmethod decorator.

A Static Method is a method that does not depend on instance variables or class variables. It is created using the @staticmethod decorator and is generally used for utility functions.

```python
class Movie:

    # Class Variable
    theater_name = "Cineplex"

    def __init__(self, title, ticket_price):
        self.title = title
        self.ticket_price = ticket_price

    # Instance Method
    def movie_info(self):
        print(f"Movie: {self.title}")
        print(f"Ticket Price: ${self.ticket_price}")

    # Class Method
    @classmethod
    def theater_info(cls):
        print(f"Theater Name: {cls.theater_name}")

    # Static Method
    @staticmethod
    def greeting():
        print("Welcome to the Movie Theater!")


movie1 = Movie("Inception", 12)
movie2 = Movie("Avatar", 15)

movie1.movie_info()
movie2.movie_info()

Movie.theater_info()

Movie.greeting()
```

## Decorators
A Decorator is a special function in Python that adds extra functionality to another function without changing its original code.

A decorator wraps around another function and performs additional tasks before or after that function executes.

## Why do we use Decorators?
Decorators allow us to add extra functionality such as:

Logging

Authentication

Timing functions

Printing messages

Error handling

without modifying the original function.


```python
# ==========================================
# Decorator Function
# ==========================================

def report_decorator(func):

    def wrapper():
        print("Starting...")

        func()

        print("Finished!")

    return wrapper


# ==========================================
# Original Function
# ==========================================

@report_decorator
def display_report():
    print("Displaying Sales Report")


display_report()
```

```python
def logger(func):

    def wrapper(*args, **kwargs):
        print("Starting Function...")

        result = func(*args, **kwargs)

        print("Function Finished.")

        return result

    return wrapper


@logger
def multiply(a, b):
    return a * b


answer = multiply(5, 4)

print(answer)
```

## Dataclass

A Dataclass is a special type of class introduced in Python that is mainly used to store data.

Instead of writing repetitive code like constructors (__init__()), Python automatically creates it for us.

A Dataclass automatically creates common methods such as __init__(), __repr__(), and __eq__() so that we can write less code.

## Why do we use Dataclasses?

Imagine you are creating many classes like:
  - Student
  - Book
  - Employee
  - Product

Each class needs:

Variables
Constructor (__init__())

Writing the constructor every time becomes repetitive.

Dataclasses automatically generate these methods.




```python
from dataclasses import dataclass


@dataclass
class Student:

    name: str
    age: int
    major: str


student1 = Student("Shristi", 30, "Marketing Analytics")
student2 = Student("John", 25, "Computer Science")

print(student1)
print(student2)
```

## Mini Project

Build a Student Management System.

Requirements:
- Add students
- Store marks
- Display details
- Use encapsulation
- Use one class variable
- Include one static method
- Include one class method


```python

class Student:

    # Class variable
    school_name = "Wright State University"

    def __init__(self, student_id, name):
        # Public instance variables
        self.student_id = student_id
        self.name = name

        # Private instance variable for encapsulation
        self.__marks = []


    # Instance method to add marks
    def add_mark(self, mark):

        if Student.is_valid_mark(mark):
            self.__marks.append(mark)
            print(f"{mark} added for {self.name}.")

        else:
            print("Invalid mark. Enter a value between 0 and 100.")


    # Instance method to safely access marks
    def get_marks(self):
        return self.__marks.copy()


    # Instance method to calculate the average
    def calculate_average(self):

        if len(self.__marks) == 0:
            return 0

        return sum(self.__marks) / len(self.__marks)


    # Instance method to display student details
    def display_details(self):

        print("\n------------------------------")
        print("Student Details")
        print("------------------------------")
        print(f"School: {Student.school_name}")
        print(f"Student ID: {self.student_id}")
        print(f"Name: {self.name}")
        print(f"Marks: {self.__marks}")
        print(f"Average: {self.calculate_average():.2f}")
        print(f"Result: {self.get_result()}")


    # Instance method to determine the result
    def get_result(self):

        if len(self.__marks) == 0:
            return "No marks available"

        if self.calculate_average() >= 50:
            return "Pass"

        return "Fail"


    # Static method
    @staticmethod
    def is_valid_mark(mark):
        return 0 <= mark <= 100


    # Class method
    @classmethod
    def change_school_name(cls, new_name):
        cls.school_name = new_name
        print(f"School name changed to {cls.school_name}.")


# ==========================================================
# Student Management System
# ==========================================================

students = []


# Function to add a student to the system
def add_student(student_id, name):

    student = Student(student_id, name)

    students.append(student)

    print(f"\n{name} was added successfully.")

    return student


# ==========================================================
# Adding Students
# ==========================================================

student1 = add_student(101, "Shristi")
student2 = add_student(102, "John")


# ==========================================================
# Adding Marks
# ==========================================================

student1.add_mark(85)
student1.add_mark(90)
student1.add_mark(78)

student2.add_mark(65)
student2.add_mark(45)
student2.add_mark(55)


# ==========================================================
# Displaying Student Details
# ==========================================================

student1.display_details()
student2.display_details()


# ==========================================================
# Changing the Class Variable Using the Class Method
# ==========================================================

Student.change_school_name("ABC University")


# Both students now display the new school name
student1.display_details()
student2.display_details()
```
