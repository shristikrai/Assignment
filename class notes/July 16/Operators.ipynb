```python
# ==========================================================
# TOPIC: OPERATORS
# ==========================================================

# Operators are symbols or words that perform operations on
# values and variables.

# ==========================================================
# 1. ARITHMETIC OPERATORS
# ==========================================================

a = 10
b = 3

print(a + b)   # Addition
print(a - b)   # Subtraction
print(a * b)   # Multiplication
print(a / b)   # Regular division
print(a // b)  # Floor division
print(a % b)   # Modulus or remainder
print(a ** b)  # Exponent or power
```

```python

# ==========================================================
# 2. COMPARISON OPERATORS
# ==========================================================
#comparison operators always produce True or False

first_number = 10
second_number = 20

print(first_number == second_number)  # Equal to
print(first_number != second_number)  # Not equal to
print(first_number > second_number)   # Greater than
print(first_number < second_number)   # Less than
print(first_number >= second_number)  # Greater than or equal to
print(first_number <= second_number)  # Less than or equal to

```

```python
# ==========================================================
# 3. LOGICAL OPERATORS
# ==========================================================

age = 25

print(age >= 18 and age < 30) #and:Both conditions must be True.
print(age < 18 or age > 60) #or:At least one condition must be True.
print(not age == 25) #not:Reverses the result.
```

```python
# ==========================================================
# 4. ASSIGNMENT OPERATORS
# ==========================================================

score = 50

score += 10 # score = score + 10
print(score)

score -= 5 #score = score - 5
print(score)

score *= 2 #score = score * 2
print(score)

score /= 5 #score = score / 5
print(score)

#here each assignment operator updates the value of score before the next line runs.
#the value of second score would be the final output of first assignment operator, i.e 60
```
