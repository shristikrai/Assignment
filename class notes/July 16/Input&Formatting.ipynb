```python
# ==========================================================
# TOPIC: INPUT AND STRING FORMATTING
# ==========================================================

# ==========================================================
# INPUT FUNCTION
# ==========================================================

# input() allows a user to enter information while the
# program is running.
#
# Important:
# input() always returns a string.

name = input("Enter your name: ")

print("Hello", name)
print(type(name))

# Converting numeric input

age = int(input("Enter your age: "))

print("Next year you will be", age + 1)
```

```python
# ==========================================================
# F-STRINGS
# ==========================================================

# An f-string allows us to insert variables or expressions
# directly inside a string.
#
# Add the letter f before the quotation mark.
# Place variables inside curly brackets.

student_name = "Shristi"
student_age = 50

print(f"My name is {student_name}.")
print(f"I am {student_age} years old.")


price = 20
quantity = 3
print(f"Total price: ${price * quantity}")
```

```python
# ==========================================================
# .format() METHOD
# ==========================================================

# The .format() method also inserts values into strings.
# Empty curly brackets act as placeholders.

print("My name is {}.".format(student_name))

print(
    "My name is {} and I am {} years old.".format(
        student_name,
        student_age
    )
)


# ==========================================================
# PRACTICE
# ==========================================================

city = input("Enter your city: ")
major = input("Enter your major: ")

print(f"You live in {city} and study {major}.")
```
