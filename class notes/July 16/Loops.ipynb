```python
# ==========================================================
# TOPIC: FOR LOOPS
# ==========================================================
# A loop repeats a block of code.
#
# A for loop is used to visit each item in a sequence or
# repeat code a known number of times.

# ==========================================================
# 1. FOR LOOP WITH RANGE
# ==========================================================

# range(5) produces:
# 0, 1, 2, 3, 4
#
# The ending number is not included.

for number in range(5):
    print(number)
```

```python
# ==========================================================
# 2. RANGE WITH START AND STOP
# ==========================================================

for number in range(2, 8): #range(start,stop) - so here it will start from 2 and end on 7
    print(number)
```

```python
#==========================================================
# 3. RANGE WITH STEP
# ==========================================================

for number in range(2, 20, 3): # The third value is called the step.It tells Python how much to increase during each repetition.
    print(number)
```

```python
# ==========================================================
# 4. LOOPING THROUGH A LIST
# ==========================================================

fruits = ["Apple", "Banana", "Mango"]

for fruit in fruits:
    print(fruit)

# During each repetition, fruit stores one item from the list.
```

```python
# ==========================================================
# 5. LOOPING THROUGH A STRING
# ==========================================================

for letter in "Python":
    print(letter)

# A string is a sequence of characters.
# Therefore, the loop processes one character at a time.
```

```python

student_names = ["Shristi", "Maya", "Riya", "Sara"]

for student in student_names:
    print(f"Welcome, {student}!")
```
