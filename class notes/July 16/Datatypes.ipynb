```python
# ==========================================================
# TOPIC: BASIC DATA TYPES
# ==========================================================

# A data type tells Python what kind of value is stored in a
# variable.
#
# The data type also determines which operations can be
# performed on that value.

# ==========================================================
# INTEGER
# ==========================================================

# An integer stores whole numbers without decimals.

age = 30
temperature_change = -5

print(age)
print(type(age))

print(temperature_change)
print(type(temperature_change))


```

```python
# ==========================================================
# FLOAT
# ==========================================================

# A float stores numbers containing decimal points.

product_price = 19.99
temperature = 72.5

print(product_price)
print(type(product_price))

print(temperature)
print(type(temperature))


```

```python
# ==========================================================
# STRING
# ==========================================================

# A string stores text, letters, numbers, or symbols inside
# quotation marks.

student_name = "Shristi"
message = "Welcome to Python"
student_code = "12345"

print(student_name)
print(type(student_name))

print(message)
print(type(message))

print(student_code)
print(type(student_code))

# Even though "12345" contains numbers, Python treats it as a
# string because it is inside quotation marks.
```

```python

# ==========================================================
# BOOLEAN
# ==========================================================

# A Boolean stores only True or False.

is_student = True
is_logged_in = False

print(is_student)
print(type(is_student))

print(is_logged_in)
print(type(is_logged_in))


```

```python
# ==========================================================
# COMPLEX NUMBER
# ==========================================================

# A complex number contains a real part and an imaginary part.
# Python uses the letter j for the imaginary part.

complex_number = 3 + 4j

print(complex_number)
print(type(complex_number))


```

```python
# ==========================================================
# PRACTICE
# ==========================================================

name = "Shristi"
age = 30
gpa = 3.8
graduated = True

print(name, type(name))
print(age, type(age))
print(gpa, type(gpa))
print(graduated, type(graduated))
```
