```python
# ==========================================================
# TOPIC: TYPE CASTING
# ==========================================================

# ----------------------------------------------------------
# What is type casting?
# ----------------------------------------------------------
# Type casting means converting a value from one data type
# into another data type.
#
# Type casting is useful when a value is not in the correct
# format for the operation we want to perform.
#
# Python has two types of type casting:
#
# 1. Implicit type casting
# 2. Explicit type casting
# ----------------------------------------------------------

# ==========================================================
# 1. IMPLICIT TYPE CASTING
# ==========================================================

# In Implicit Type Casting Python automatically converts one data type into another.

integer_number = 10 #this variable contains an integer
decimal_number = 2.5 #this variable contains a float

result = integer_number + decimal_number
# Python automatically converts 10 into 10.0 before performing
# the addition.

print(result)
print(type(result))

# The final result is therefore a float.
```

```python
# ==========================================================
# 2. EXPLICIT TYPE CASTING
# ==========================================================

# Explicit type casting means that the programmer manually
# converts the value using functions such as:
#
# int()
# float()
# str()
# bool()

# String to integer

age_text = "30"
age_number = int(age_text)

print(age_number)
print(type(age_number))
print(age_number + 5)
```

```python
# Float to integer

height = 5.9
height_integer = int(height)

print(height_integer)
print(type(height_integer))

# Important:
# int() does not round the number.
# It removes the decimal part.
#
# 5.9 becomes 5.
```

```python
# Integer to float

marks = 85
marks_float = float(marks)

print(marks_float)
print(type(marks_float))
```

```python

# Integer to string

phone_code = 937
phone_code_string = str(phone_code)

print(phone_code_string)
print(type(phone_code_string))
```

```python
# Boolean conversion

print(bool(1)) #The bool() function checks whether a value is considered True or False.
print(bool(0))
print(bool("Python")) #non empty string
print(bool("")) #empty string

#Any non-empty string is True and empty is False in Python
```
